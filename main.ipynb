```python
EPOCH = 150
BIT = 32
BACKBONE = "ViT-B_32"
DATASET = "cifar10"

SAVE_CSQ  = f"Checkpoints_Results/CSQ-{BACKBONE}-{DATASET}"
SAVE_DPN  = f"Checkpoints_Results/DPN-{BACKBONE}-{DATASET}"
SAVE_DSH  = f"Checkpoints_Results/DSH-{BACKBONE}-{DATASET}"
SAVE_GH   = f"Checkpoints_Results/GreedyHash-{BACKBONE}-{DATASET}"
SAVE_HN   = f"Checkpoints_Results/HashNet-{BACKBONE}-{DATASET}"
SAVE_IDHN = f"Checkpoints_Results/IDHN-{BACKBONE}-{DATASET}"
```

```python
import subprocess

SMOKE_EPOCH = 10
SMOKE_TEST_MAP = 5

frameworks = [
    ("CSQ",        "CSQ.py",        SAVE_CSQ),
    ("DPN",        "DPN.py",        SAVE_DPN),
    ("DSH",        "DSH.py",        SAVE_DSH),
    ("GreedyHash", "GreedyHash.py", SAVE_GH),
    ("HashNet",    "HashNet.py",    SAVE_HN),
    ("IDHN",       "IDHN.py",       SAVE_IDHN),
]

results = []
for name, script, save_path in frameworks:
    print(f"=== SMOKE {name} ===")
    cmd = ["python", script,
           "--dataset", DATASET, "--bit", str(BIT),
           "--epoch", str(SMOKE_EPOCH), "--test_map", str(SMOKE_TEST_MAP),
           "--save_path", save_path]
    rc = subprocess.call(cmd)
    results.append((name, "OK" if rc == 0 else f"FAIL(rc={rc})"))

print("\n=== SUMMARY ===")
for name, status in results:
    print(f"{name}: {status}")
```
